## Missing Value Handling

### 1. Description Column

The `Description` column contained 1,454 missing values (approximately 0.27% of the total records).

To handle these missing values, a StockCode-based mode imputation technique was applied.

**Methodology:**
- Records were grouped based on their `StockCode`.
- The most frequently occurring product description (mode) was identified within each group.
- Missing descriptions were replaced with the corresponding mode.
- If no valid description was available for a particular StockCode, the missing value was replaced with "Unknown".

**Python Code:**

```python
df['Description'] = df.groupby('StockCode')['Description'].transform(
    lambda x: x.fillna(
        x.mode().iloc[0] if not x.mode().empty else 'Unknown'
    )
)
```

**Reason for Selection:**

StockCode represents a product identifier. Therefore, using the most frequent description associated with the same StockCode is a suitable approach for recovering missing product descriptions.

### 2. CustomerID Column

The `CustomerID` column contained 135,080 missing values (approximately 24.93% of the dataset).

**Treatment:**
- Missing CustomerID values were retained as NaN.
- Mean and median imputation were not applied because CustomerID is an identifier rather than a numerical measurement.
- Artificial customer IDs were not generated to avoid creating incorrect customer identities.
- Records with missing CustomerID were retained for overall sales analysis and forecasting.

**Reason for Selection:**

Removing these records would result in losing approximately 25% of the transaction data. Since these transactions still contain useful information such as Quantity, UnitPrice and InvoiceDate, they can contribute to sales forecasting.

### 3. Conclusion

Missing values were handled according to the business meaning of each column. Description values were imputed using StockCode-based mode imputation, while missing CustomerID values were retained as NaN.

This approach helps preserve transaction information and maintain data quality for further exploratory data analysis, preprocessing and future sales prediction.

### Handling Missing Values in the Description Column

**Objective:** Fill missing values in the `Description` column using the most frequently occurring description for each `StockCode`.

**Code:**

```python
df['Description'] = df.groupby('StockCode')['Description'].transform(
    lambda x: x.fillna(
        x.mode().iloc[0] if not x.mode().empty else 'Unknown'
    )
)
```

**Step-by-Step Explanation:**

1. `groupby('StockCode')`: Groups records based on unique StockCode values.
2. `['Description']`: Selects the Description column from each group.
3. `transform()`: Applies the function to each group while maintaining the original DataFrame structure.
4. `lambda x`: Represents the Description values of each StockCode group.
5. `x.mode()`: Finds the most frequently occurring description.
6. `iloc[0]`: Selects the first mode if multiple modes exist.
7. `if not x.mode().empty else 'Unknown'`: Uses the first mode if available; otherwise, fills with 'Unknown'.
8. `fillna()`: Replaces missing Description values with the selected value.

**Example:**

Before Imputation:

| StockCode | Description |
|---|---|
| 1001 | Laptop |
| 1001 | Laptop |
| 1001 | NaN |
| 1002 | Mouse |
| 1002 | NaN |
| 1002 | Mouse |

After Imputation:

| StockCode | Description |
|---|---|
| 1001 | Laptop |
| 1001 | Laptop |
| 1001 | Laptop |
| 1002 | Mouse |
| 1002 | Mouse |
| 1002 | Mouse |

**Why did I use this method?**

- The same StockCode generally represents the same product.
- Missing descriptions can be recovered from other transactions with the same StockCode.
- Group-based mode imputation is more meaningful than filling all missing values with a generic value.
- It helps preserve useful product information.

**Conclusion:**

I used group-based mode imputation to handle missing values in the Description column. This method fills missing product descriptions using the most frequently occurring description associated with the same StockCode, maintaining data consistency.
